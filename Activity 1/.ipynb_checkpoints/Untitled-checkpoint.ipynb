```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img = Image.open('hog.png')  # from PIL (Pillow) module 

GS = np.array(img.convert('L'))

flat_GS = GS.flatten()
PDF_hist, PDF_edges = np.histogram(GS, bins = 256, range = (0,256), density=True)
CDF = np.cumsum(PDF_hist) 


def sigmoid(x):
    return 1 / (1 + np.exp(-x))

domain = np.linspace(-10, 10, 256)
CDF_sigmoid = sigmoid(domain)

# plt.plot(CDF_sigmoid)
# print(CDF_sigmoid.min())
x = np.arange(0, 256)

new_flatGS_2 = np.interp(CDF[flat_GS], domain, CDF_sigmoid)
new_GS_2 = np.reshape(new_flatGS_2, np.shape(GS))

PDF_hist_new_2, PDF_edges_new_2 = np.histogram(new_GS_2, bins = 256 ,density=True) 
CDF_new_2 = np.cumsum(PDF_hist_new_2) 

# plt.plot(CDF_new_2)

plt.imshow(new_GS_2, cmap = 'gray')
```

````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import cv2
import pandas as pd 

from scipy.ndimage import binary_dilation, binary_erosion
from skimage.morphology import opening, closing, area_closing, area_opening
from skimage.morphology import star, disk

from skimage.measure import label, regionprops, regionprops_table
````

````python
# source: https://medium.com/swlh/image-processing-with-python-connected-components-and-region-labeling-3eef1864b951
````

````python
img = Image.open('Inputs/sample2.jpg')
img_array = np.array(img)
sample_GS = np.array(img.convert('L'))

threshold = 110
sample_segmented_binary =  np.invert(cv2.threshold(sample_GS, threshold, 255, cv2.THRESH_BINARY)[1])
# plt.imshow(sample_segmented_binary, cmap='gray')

# We clean up with morphological 
strel = disk(8)

sample_closed = closing(sample_segmented_binary, footprint=strel)
# sample_opened_opened = opening(sample_opened, footprint=strel)

# plt.imshow(sample_opened, cmap='gray_r')


fig, ax = plt.subplots(1,3)

ax[0].imshow(img)
ax[1].imshow(sample_segmented_binary, cmap='gray')
ax[2].imshow(sample_closed, cmap='gray')

for i in range(3):
    ax[i].axis('off')
ax[0].set_title('Sample Image')
ax[1].set_title('Thresholded')
ax[2].set_title('Morphology-Treated')
    
fig.savefig('Results/Segmented.png',dpi=500)
# plt.imshow(sample_opened, cmap='gray_r')
# plt.imshow(sample_opened_closed, cmap='gray_r')
````

````python
label_im = label(sample_closed)# label pixels based on connectivity. count starts at upper right region
regions = regionprops(label_im)# measure properties of labeled objects, use regionprops_table to show

plt.title('Labeled Image')
plt.axis('off')
plt.savefig('Results/Labeled.png', dpi=500)
plt.imshow(label_im)
````

````python
len(regions)   # skimage measures 130 parameters!!!
````

````python
properties = ['area','bbox_area', 'extent',  
              'mean_intensity', 'solidity', 'eccentricity', 
              'orientation']
measurement_table = pd.DataFrame(regionprops_table(label_im, sample_GS, 
             properties=properties))
````

After the top right, the label function labels regions from left to right and top to bottom

Notice that some of the areas are very negligible, this can be attributed to noise which we will remove. We filter out these regions with standard pandas dataframe manipulation. 

````python
masks = []
bbox = []
list_of_index = []
for num, x in enumerate(regions):
    area = x.area   
    convex_area = x.convex_area
    if num!=0 and (area>6000):    # filter out entries with area less than 6000, manually chosen
                                  # add more (and ____) to filter out more parameters
        masks.append(regions[num].convex_image)
        bbox.append(regions[num].bbox)   
        list_of_index.append(num)
count = len(masks)

# len(list_of_index)  # index of accepted objects
# plt.imshow(masks[0])  # a sample mask, which is an binary image of an object
# print(bbox)           # bbox of the objects
````

We use the bounding box parameter of ```regionprops``` to determine the object's coordinates in the image. Then, we slice the original image to this bounding box to get an image of each region. 

We also create a mask that hides unwanted regions in the original image. We use the ```label ``` array of the desired regions, and using the mask with the original image. 

````python
fig, ax = plt.subplots(2, int(count/2), figsize=(15,3))
for axis, box, mask in zip(ax.flatten(), bbox, masks):
    red  =  img_array[:,:,0][box[0]:box[2], box[1]:box[3]] * mask
    green = img_array[:,:,1][box[0]:box[2], box[1]:box[3]] * mask
    blue  = img_array[:,:,2][box[0]:box[2], box[1]:box[3]] * mask
    image = np.dstack([red,green,blue])
    axis.imshow(image)
    axis.axis('off')    # i can only display multiples of 2. In this case there are 35 objects but
                        # i could only show 34. Nevertheless, they are still accesible in good_measurements
````

Lettuce display the properties of the accepted regions / blobs of interest. 

````python
good_measurements = measurement_table.loc[list_of_index]
good_measurements
````

````python
rgb_mask = np.zeros_like(label_im)
for x in list_of_index:
    rgb_mask += (label_im==x+1).astype(int)
red  =  img_array[:,:,0] * rgb_mask
green = img_array[:,:,1] * rgb_mask
blue  = img_array[:,:,2] * rgb_mask
image = np.dstack([red, green, blue])
plt.title('Labeled and Segmented Image')
plt.axis('off')
plt.imshow(image)
plt.savefig('Results/SegmentedRGB.png', dpi=500)
````

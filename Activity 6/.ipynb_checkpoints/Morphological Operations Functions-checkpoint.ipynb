```python
import numpy as np
from scipy.ndimage import binary_dilation, binary_erosion
import matplotlib.pyplot as plt
```

```python
square_5 = np.zeros((11,11))
square_5[3:8, 3:8] = 1               # 5x5 square

plus_sign = np.zeros((11, 11))
plus_sign[3:8, 5] = 1
plus_sign[5, 3:8] = 1               # plus sign - 1 box thick

donut = np.zeros((20, 20))
donut[5:15, 5:15] = 1
donut[7:13, 7:13] = 0               # 10 X 10 with hole in middle

dumbell = np.zeros((21, 21))
dumbell[8:13, 4:9] = 1
dumbell[8:13, 12:17] = 1
dumbell[10, 9:12] = 1               # dumbell

# ==================================================================
kernel_square_5 = np.zeros((11,11))
kernel_square_5[3:8, 3:8] = 1                       # 5x5 square

kernel_square_2 = np.zeros((10,10))
kernel_square_2[4:6, 4:6] = 1                       # 2 X 2 square

kernel_rectangle_21 = np.zeros((10,10))
kernel_rectangle_21[5:6, 4:6] = 1                   # 2 X 1 rectangle
```

```python
fig, ax = plt.subplots(3,4)

for row in range(3):
    ax[row,0].imshow(square_5, cmap='gray')   # image
    
ax[0,1].imshow(kernel_square_2, cmap='gray')
ax[1,1].imshow(kernel_rectangle_21, cmap='gray')
ax[2,1].imshow(kernel_square_5, cmap='gray')       # kernels

ax[0,2].imshow(binary_dilation(square_5, structure = kernel_square_2, origin = (4,4)), cmap='gray')
ax[1,2].imshow(binary_dilation(square_5, structure = kernel_rectangle_21, origin = (5,4)), cmap='gray')
ax[2,2].imshow(binary_dilation(square_5, structure = kernel_square_5, origin = (5,5)), cmap='gray')       # binary_dilations

ax[0,3].imshow(erosion(square_5, structure = kernel_square_2, origin = (4,4)), cmap='gray')
ax[1,3].imshow(erosion(square_5, structure = kernel_rectangle_21, origin = (5,4)), cmap='gray')
ax[2,3].imshow(erosion(square_5, structure = kernel_square_5, origin = (5,5)), cmap='gray')    # erosions


ax[0,0].set_title('Image')
ax[0,1].set_title('Kernel')
ax[0,2].set_title('binary_dilation')
ax[0,3].set_title('Erosion')
    

for j in range(4):
    for i in range(3):
        ax[i, j].axis('off')
    
fig.savefig('Results/5X5 Square Image')
```
